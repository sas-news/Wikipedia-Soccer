# Wikipedia Soccer - Google Colab Port

このノートブックは、React + Express (TypeScript) で構築された Wikipedia Soccer を Google Colab 上で直接実行するためのものです。
Colab には標準で Node.js がインストールされているため、ソースコードを Python に書き直すことなく、バックグラウンドで Express サーバーを立ち上げ、Colab の iframe 機能を使って UI を表示できます。

```python
# 1. プロジェクトの準備
# Colabにフォルダごとアップロードするか、GitHubからクローンしてディレクトリを移動します。
# 例: !git clone https://github.com/your-repo/wikipedia-soccer.git
# %cd wikipedia-soccer

# パッケージのインストール
!npm install
```

```python
import subprocess
import time
from google.colab import output

PORT = 3000

# 2. Expressサーバー(server.ts)をバックグラウンドで起動
print("バックエンドサーバーを起動中 (tsx server.ts)...")
# stdout/stderrをPIPEにしてColabのセル出力を汚さないようにします
server_process = subprocess.Popen(
    ["npx", "tsx", "server.ts"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

# サーバーの立ち上がりを少し待つ
time.sleep(5)
print(f"ポート {PORT} でサーバーが起動しました。iframeを展開します...")

# 3. 起動したサーバーの画面をColab上に埋め込み表示
output.serve_kernel_port_as_iframe(PORT, height=800)
```

```python
# サーバーを停止する場合（再起動したい時など）
server_process.terminate()
print("サーバーを停止しました。")
```
